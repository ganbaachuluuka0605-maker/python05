```python
from abc import ABC, abstractmethod

class Account(ABC):                      # Abstraction
    bank_name = "MUBIS Bank"             # Class attribute (бүх объектод нийтлэг)

    def __init__(self, owner, balance=0):    # Constructor
        self.owner = owner               # Instance attribute
        self.__balance = balance         # Encapsulation (private)

    @property
    def balance(self):                   # Зөвхөн уншигдах шинж
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Дүн эерэг байх ёстой")
        self.__balance += amount

    def _withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Үлдэгдэл хүрэлцэхгүй")
        self.__balance -= amount

    @abstractmethod
    def monthly_update(self): ...        # Дэд класс заавал хэрэгжүүлнэ

    def __str__(self):                   # Dunder (magic) method
        return f"{self.owner}: {self.balance:,.0f}₮"


class Savings(Account):                  # Inheritance
    def __init__(self, owner, balance=0, rate=0.01):
        super().__init__(owner, balance) # Эх классын constructor
        self.rate = rate

    def monthly_update(self):            # Polymorphism: хүү нэмнэ
        self.deposit(self.balance * self.rate)


class Checking(Account):
    FEE = 1000

    def monthly_update(self):            # Polymorphism: шимтгэл хасна
        self._withdraw(self.FEE)


accounts = [Savings("Бат", 100_000, 0.02), Checking("Сараа", 50_000)]
for acc in accounts:
    acc.monthly_update()                 # Ижил дуудлага, өөр үйлдэл
    print(acc)
# Бат: 102,000₮
# Сараа: 49,000₮

# acc.__balance = 0   → AttributeError биш ч жинхэнэ үлдэгдлийг өөрчлөхгүй
# Account("X")        → TypeError: abstract класс тул объект үүсгэх боломжгүй
```

```python
input_data = """Уламбаяр Төртогтох эрэгтэй МБУС МЗТ 99044079
Баттогтох Цэен-Ойдов эрэгтэй МБУС МЗТ 99003771
Сандагдорж Балдорж эрэгтэй МБУС МТ 88859959
Уртнасан Гантогтох эмэгтэй МБУС МЗТ 88081699
Алимаа Бор эмэгтэй НХУС ТТ 88169994
Тунгалаг Балжир эмэгтэй НХУС АХТ 80159991"""

with open("input.txt", "w", encoding="utf-8") as f:
    f.write(input_data)
# input.txt файлыг utf-8 өгөгдөл уншихаар нээж байна.
with open("input.txt", "r", encoding="utf-8") as f:
    # Оролтын файлын мөр бүрийг уншиж mylines list рүү хадгалж байна.
    mylines = f.readlines()

# Багш нарын мэдээлэл хадгалах List үүсгэж байна.
teachers = []

# Салбар бүрийн багшийг тоолох Dict үүсгэж байна.
salbar = {}

# mylines list-ээр давталт гүйлж мөр бүрийг баганаар салгаж rd гэдэг Dict үүсгэж teachers list-рүү нэмж байна.
for line in mylines:
    line = line.strip()
    if not line:
        continue
    parts = line.split()
    if len(parts) >= 6:
        rd = {
            "ner": parts[0],
            "ovog": parts[1],
            "huis": parts[2],
            "salbar": parts[3],
            "tenhim": parts[4],
            "utas": parts[5],
        }
        teachers.append(rd)

        # Хэрэв rd['salbar'] утгатай key үүссэн бол тэр салбарын багшийн тоог 1-ээр нэмж,
        # үүсээгүй бол шинээр үүсгэн анхны 1 гэсэн утга оноож байна.
        s_name = rd["salbar"]
        if s_name in salbar:
            salbar[s_name] += 1
        else:
            salbar[s_name] = 1

# Гаралтын output.txt файлыг utf-8 өгөгдөл бичихээр нээж байна.
with open("output.txt", "w", encoding="utf-8") as f_out:
    # salbar dict объектийг key-р нь өсөхөөр эрэмбэлж эрэмбийн дагуу давталт гүйлгэн багш нарын тоог хэвлэж байна.
    for s_name in sorted(salbar.keys()):
        f_out.write(f"{s_name} {salbar[s_name]}\n")

# Үр дүнг дэлгэцэнд харуулах
print("--- output.txt файлын агуулга ---")
with open("output.txt", "r", encoding="utf-8") as f_out:
    print(f_out.read())
```
